# Personal Expense Tracker

A command-line app for logging daily expenses, grouping them by category, and comparing totals against a monthly budget. Expenses persist in a CSV file so previous entries reload when the program starts.

The menu lets you add an expense (date, category, amount, description), view the list with a check for incomplete rows, set or check the monthly budget, save to file, or save and exit.

```python
import pandas as pd
```

```python
def add_expense():
    global df

    # Get the new data
    date = input('Enter the date in YYYY-MM-DD: ')
    category = input('Enter a category: ').lower()
    amount = float(input('Enter an amount: '))
    description = input('Enter a description: ')
    
    # Add a new row of data
    dictionary = {
        'date': [date], 
        'category': [category], 
        'amount': [amount], 
        'description': [description]
    }
    new_row = pd.DataFrame(dictionary)
    if df.empty:
        df = new_row
    else:
        df = pd.concat([df, new_row], ignore_index=True)
```

```python
def view_expenses():
    if len(df) == 0:
        print('You have not added any expenses yet. Enter 1 to add an expense')
    else:
        print('Your current expenses:')
        print(f"{'Date':<10} | {'Category':<12} | {'Amount':<9} | {'Description'}")  
        print('-' * 52)
        for i, row in df.iterrows():
            if (not row['date']) or not (row['category']) or (row['amount'] <= 0) or (not row['description']):
                print('< missing data for this row')
            print(f"{row['date']:<10} | {row['category']:<12} | ${row['amount']:>10,.2f} | {row['description']}")
```

```python
def set_budget():
    new_budget = float(input('Set a new budget: '))
    budget_file = open('budget.txt', 'w')
    budget_file.write(str(new_budget))
    budget_file.close()
```

```python
def track_budget():
    try:
        budget_file = open('budget.txt', 'r')
        budget = budget_file.read()
    except:
        budget_file = open('budget.txt', 'w')
        budget = ''
    
    spent = df['amount'].sum()
    
    if len(budget) == 0:
        print(f'You have not set a budget yet.\nYou have spent ${spent:,.2f}.')
    else:
        budget = float(budget)
        print(f'Your current budget is ${budget:,.2f}.')
        
        if spent < budget:
            print(f'You have spent ${spent:,.2f} this month.')
            print(f'You have ${(budget - spent):,.2f} remaining for the month.')
        elif spent == budget:
            print('You have filled your quota for the month')
        elif spent > budget:
            print(f'⚠️ You have exceeded your budget for this month by ${spent - budget:,.2f}! ⚠️')
        
    budget_file.close()
```

```python
def save():
    df.to_csv('expenses.csv', index=False)
```

# Interactive Menu

```python
# Import files
try:
    df = pd.read_csv('expenses.csv')
except:
    df = pd.DataFrame(columns=['date', 'category', 'amount', 'description'])
    df.to_csv('expenses.csv', index=False)

    
# Input loop
while True:
    option = input("""\n\nPlease select an option:
    1. Add an expense
    2. View expenses
    3. Set your budget
    4. Track your budget
    5. Save expenses to a file
    6. Save and exit

    --> """)

    print()

    if not option.strip():
        print('Saving to the file and exiting...')
        save()
        break
    
    match int(option):
        case 1: 
            add_expense()

        case 2:
            view_expenses()

        case 3:
            set_budget()
        
        case 4:
            track_budget()

        case 5:
            print('Saving to the file...')
            save()
            
        case 6:
            print('Saving to the file and exiting...')
            save()
            break
            
        case _:
            print('Invalid choice. Please enter a single number from the list. Enter 6 to end.')
```
